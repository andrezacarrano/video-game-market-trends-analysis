## Padronização dos nomes das colunas

Renomeamos todas as colunas para letras minúsculas e removemos espaços em branco para facilitar o uso no código e evitar erros de digitação.


```python
#Alterando as colunas para minúsculo
df.columns = df.columns.str.lower().str.strip()
```

```python
#Convertendo os dados year_of_release: e user_score: para float e alterando o TDB para NaN
df['user_score'] = pd.to_numeric(df['user_score'].replace('tdb',np.nan), errors = 'coerce')
df['year_of_release'] = pd.to_numeric(df['year_of_release'], errors = 'coerce')
```

### Valores ausentes

Durante a inspeção inicial dos dados verifiquei diversas colunas com valores ausentes:

-Year_of_release = optei por excluir os valores ausentes nessa coluna, pois sem essa informação, não é possível realizar análises por período e depois transformar esse valor em número inteiro.

-Critic_score e user_scores = optei por manter os valores ausentes como NaN, pois representam avaliações e são importantes para análises como correlação com vendas e não são obrigatórias para todas as análises, permitindo que sejam ignorados apenas nas análises que exigem essas métricas.

-rating = mantive os valores ausentes como 'Unlnown' pois é uma classificação que varia por região e nem todos os jogos tem, e tratei separadamente os casos com a sigla TDB (subsituindo por NaN)

e por fim crio a coluna com o total de vendas em todas as regiões, somando por linhas.

```python
# Criar cópia do DataFrame com ano preenchido
df_clean = df.dropna(subset = ['year_of_release']).copy()
# Garantir que o ano seja inteiro
df_clean['year_of_release'] = df_clean['year_of_release'].astype(int)
# Preencher valores ausentes de 'rating' com 'Unknown'
df_clean['rating'] = df_clean['rating'].fillna('Unknown')
# Calcular total de vendas
df_clean['total_sales'] = df_clean[['na_sales', 'eu_sales', 'jp_sales', 'other_sales']].sum(axis=1)
# Preencher anos ausentes com média da plataforma (arredondada)
platform_year_mean = df_clean.groupby('platform')['year_of_release'].mean().round()
# Preencher os nulos no df original
df['year_of_release'] = df.apply(lambda row: platform_year_mean[row['platform']] if pd.isna(row['year_of_release']) else row['year_of_release'], axis = 1).astype(int)
print(df_clean.head())
print(df.head())
```

### Construa um gráfico de dispersão e calcule a correlação entre avaliações e vendas

Análise de como as avaliações de usuários e críticos influenciam as vendas em uma plataforma popular. 

Vamos usar a PS3 como exemplo, já que ela teve vendas consistentes entre 2005 e 2015

```python
# Selecionar jogos da plataforma PS3 com avaliações e vendas disponíveis
df_ps3 = df_10anos[
    (df_10anos['platform'] == 'PS3') &
    (df_10anos['user_score'].notna()) &
    (df_10anos['critic_score'].notna()) &
    (df_10anos['total_sales'].notna())
]

#Filtramos apenas os jogos da PS3 que têm avaliações e vendas registradas — isso evita distorções por dados ausentes.
```

```python
# Gráfico de dispersão entre nota dos usuários e vendas (user_score vs total_sales)
plt.figure(figsize=(8,5))
plt.scatter(df_ps3['user_score'],df_ps3['total_sales'],alpha=0.6)
plt.title('Relação entre nota dos usuários e vendas (PS3)')
plt.xlabel('Nota dos usuários')
plt.ylabel('Vendas globais (em milhões)')
plt.show()
```

```python
# Correlação de Pearson (linear) e Spearman (ordem)

#Pearson mede correlação linear (ex: quanto maior a nota, maior a venda)
pearson = stats.pearsonr(df_ps3['user_score'],df_ps3['total_sales'])
#Spearman mede correlação por ranking (ex: jogos com nota alta tendem a vender mais)
spearman = stats.spearmanr(df_ps3['user_score'],df_ps3['total_sales'])

pearson,spearman
```

O gráfico mostra que a maioria dos jogos tem vendas abaixo de 2 milhões, mesmo com boas notas. A correlação entre nota dos usuários e vendas é fraca — os valores de Pearson e Spearman são baixos. Isso indica que uma nota alta não garante sucesso de vendas. Outros fatores como marketing, franquia e plataforma também influenciam.

## Teste de Hipótese

```python
#Hipótese 1: Xbox One vs PC

#Hipótese nula (H₀): As médias das classificações dos usuários para Xbox One e PC são iguais.
#Hipótese alternativa (H₁): As médias das classificações dos usuários para Xbox One e PC são diferentes.

#Vamos definir α = 0.05 (5%) que é padrão para testes exploratórios
#Isso significa que aceitamos um risco de 5% de rejeitar a hipótese nula quando ela for verdadeira (erro tipo I).

# Como estamos comparando médias de duas amostras independentes, o teste ideal é o teste t de Student para duas amostras independentes

# Filtrar dados para Xbox One e PC
xbox = df_10anos[df_10anos['platform'] == 'XOne']['user_score'].dropna()
pc = df_10anos[df_10anos['platform'] == 'PC']['user_score'].dropna()

from scipy.stats import ttest_ind

# Teste t de duas amostras independentes
t_stat,p_value = ttest_ind(xbox, pc, equal_var = False) # Welch's t-test

# Exibir resultados
print(f'Estatística t: {t_stat:.3f}')
print(f'Valor-p: {p_value:.3f}')

# Definir nível de significância
alpha = 0.05

if p_value < alpha:
    print("Rejeitamos H₀: as médias são diferentes.")
else:
    print("Não rejeitamos H₀: não há evidência de diferença entre as médias.")
```

```python
#Hipótese 2: Action vs Sports

#Hipótese nula (H₀): As médias das classificações dos usuários para os gêneros Action e Sports são iguais.
#Hipótese alternativa (H₁): As médias das classificações dos usuários para os gêneros Action e Sports são diferentes.

#Vamos definir α = 0.05 (5%) que é padrão para testes exploratórios
#Isso significa que aceitamos um risco de 5% de rejeitar a hipótese nula quando ela for verdadeira (erro tipo I).

# Como estamos comparando médias de duas amostras independentes, o teste ideal é o teste t de Student para duas amostras independentes

# Filtrar dados para Action e Sports
action = df_10anos[df_10anos['genre'] == 'Action']['user_score'].dropna()
sports = df_10anos[df_10anos['genre'] == 'Sports']['user_score'].dropna()

from scipy.stats import ttest_ind

# Teste t de duas amostras independentes
t_stat,p_value = ttest_ind(action, sports, equal_var = False) # Welch's t-test

# Exibir resultados
print(f'Estatística t: {t_stat:.3f}')
print(f'Valor-p: {p_value:.3f}')

# Definir nível de significância
alpha = 0.05

if p_value < alpha:
    print("Rejeitamos H₀: as médias são diferentes.")
else:
    print("Não rejeitamos H₀: não há evidência de diferença entre as médias.")
```
